## 🟢 Question 7 — `PARTITION BY`

Using the **`Employees`** table, write a SQL query to show:

- `EmployeeName`
- `DepartmentID`
- `Salary`
- A **row number within each department**, ordered by salary from highest to lowest

Use `ROW_NUMBER()`, `PARTITION BY`, and `ORDER BY`.

**Your task:** Write the SQL query.

```python
df_8 = pd.read_sql("SELECT EmployeeName,DepartmentID,SALARY,ROW_NUMBER() OVER(PARTITION BY DepartmentID ORDER BY Salary Desc) As dept_Salary_rank FROM Employees;",con =engine)
df_8.head()
```
